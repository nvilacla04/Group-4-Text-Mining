## [total points: 4] Exercise 1: NLTK
In this exercise, we use NLTK to apply **Part-of-speech (POS) tagging**, **Named Entity Recognition (NER)**, and **Constituency parsing**. The following code snippet already performs sentence splitting and tokenization. 

```python
import nltk
from nltk.tokenize import sent_tokenize
from nltk import word_tokenize
```

```python
sentences_nltk = sent_tokenize(text)
```

```python
tokens_per_sentence = []
for sentence_nltk in sentences_nltk:
    sent_tokens = word_tokenize(sentence_nltk)
    tokens_per_sentence.append(sent_tokens)
```

We will use lists to keep track of the output of the NLP tasks. We can hence inspect the output for each task using the index of the sentence.

```python
sent_id = 1
print('SENTENCE', sentences_nltk[sent_id])
print('TOKENS', tokens_per_sentence[sent_id])
```

### [point: 1] Exercise 1a: Part-of-speech (POS) tagging
Use `nltk.pos_tag` to perform part-of-speech tagging on each sentence.

Use `print` to **show** the output in the notebook (and hence also in the exported PDF!).

```python
pos_tags_per_sentence = []
for tokens in tokens_per_sentence:
    pos_tags = nltk.pos_tag(tokens)
    pos_tags_per_sentence.append(pos_tags)
    print(pos_tags)
    print()
```

```python
print(pos_tags_per_sentence)
```

### [point: 1] Exercise 1b: Named Entity Recognition (NER)
Use `nltk.chunk.ne_chunk` to perform Named Entity Recognition (NER) on each sentence.

Use `print` to **show** the output in the notebook (and hence also in the exported PDF!).

```python
from nltk.chunk import ne_chunk

ner_tags_per_sentence = []
for pos_tags in pos_tags_per_sentence:
    ner_tree = ne_chunk(pos_tags)
    ner_tags_per_sentence.append(ner_tree)
    print(ner_tree)
    print()
```

```python
print(ner_tags_per_sentence)
```

### [points: 2] Exercise 1c: Constituency parsing
Use the `nltk.RegexpParser` to perform constituency parsing on each sentence.

Use `print` to **show** the output in the notebook (and hence also in the exported PDF!).

```python
constituent_parser = nltk.RegexpParser('''
NP: {<DT>? <JJ>* <NN>*} # NP
P: {<IN>}           # Preposition
V: {<V.*>}          # Verb
PP: {<P> <NP>}      # PP -> P NP
VP: {<V> <NP|PP>*}  # VP -> V (NP|PP)*''')
```

```python
constituency_output_per_sentence = []
for pos_tags in pos_tags_per_sentence:
    parse_tree = constituent_parser.parse(pos_tags)
    constituency_output_per_sentence.append(parse_tree)
    print(parse_tree)
    print()
```

```python
print(constituency_output_per_sentence)
```

Augment the RegexpParser so that it also detects Named Entity Phrases (NEP), e.g., that it detects *Galaxy S III* and *Ice Cream Sandwich*

```python
constituent_parser_v2 = nltk.RegexpParser('''
NEP: {<NNP>+(<CD>|<NNP>|<NN>)*}   # Named Entity Phrase: proper nouns + optional nums/nouns
NP: {<DT>? <JJ>* <NN>*}            # NP
P: {<IN>}                           # Preposition
V: {<V.*>}                          # Verb
PP: {<P> <NP>}                      # PP -> P NP
VP: {<V> <NP|PP>*}                  # VP -> V (NP|PP)*''')
```

```python
constituency_v2_output_per_sentence = []
for pos_tags in pos_tags_per_sentence:
    parse_tree_v2 = constituent_parser_v2.parse(pos_tags)
    constituency_v2_output_per_sentence.append(parse_tree_v2)
    print(parse_tree_v2)
    print()
```

```python
print(constituency_v2_output_per_sentence)
```

## [total points: 1] Exercise 2: spaCy
Use Spacy to process the same text as you analyzed with NLTK.

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
sents = list(doc.sents)

for i, sent in enumerate(sents):
    print(f"--- Sentence {i} ---")
    print(sent.text)
    print()
    for token in sent:
        print(f"  {token.text:<20} pos: {token.pos_:<8} tag: {token.tag_:<6} dep: {token.dep_}")
    ents_in_sent = [e for e in doc.ents if e.start >= sent.start and e.end <= sent.end]
    if ents_in_sent:
        print("  Named Entities:")
        for ent in ents_in_sent:
            print(f"    {ent.text:<30} {ent.label_}")
    print()
```

### [points: 2] Exercise 3b: Named Entity Recognition (NER)
* Describe differences between the output from NLTK and spaCy for Named Entity Recognition. Which one do you think performs better?

### Answer

**Differences between NLTK and spaCy NER:**

**1. Label inventory:**
NLTK uses coarse labels (ORGANIZATION, PERSON, GPE, FACILITY).
spaCy uses a richer set: ORG, PERSON, GPE, MONEY, DATE, CARDINAL, LAW, PRODUCT, etc.
This means spaCy can label "$1.05bn" as MONEY and "August" as DATE; NLTK cannot.

**2. Multi-word entities:**
spaCy processes full sentence context and correctly groups multi-word names
(e.g., "San Jose" → one GPE, "Ice Cream Sandwich" → one PRODUCT).
NLTK works token-by-token after POS tagging, so it often splits these or misses them.

**3. Accuracy on this text:**
- spaCy correctly identifies: Apple (ORG), Samsung (ORG), US (GPE),
  San Jose (GPE), California (GPE), $1.05bn (MONEY), August (DATE), iPad (PRODUCT).
- NLTK misses monetary amounts and dates entirely, and fragments names like
  "Galaxy S III" into separate chunks.

**Verdict:** spaCy performs better. Its statistical neural model handles
domain-specific names and multi-token spans far more reliably than NLTK's
rule-based chunker.


### [points: 2] Exercise 3c: Constituency/dependency parsing
Choose one sentence from the text and run constituency parsing using NLTK and dependency parsing using spaCy.
* describe briefly the difference between constituency parsing and dependency parsing
* describe differences between the output from NLTK and spaCy.

### Answer

**Constituency vs Dependency parsing:**

*Constituency parsing* groups words into nested phrases (NP, VP, PP, S) forming a
hierarchical tree. It answers: *what are the phrase-level building blocks of the sentence?*

*Dependency parsing* draws a directed arc from each word to its syntactic head,
labelled with a grammatical relation (nsubj, dobj, prep, conj, …). It answers:
*which word governs which other word, and how?*

**Chosen sentence (index 3):**
*"In August, Samsung lost a US patent case to Apple and was ordered to pay its rival
$1.05bn (£0.66bn) in damages for copying features of the iPad and iPhone in its Galaxy
range of devices."*

**NLTK constituency output:**
The RegexpParser produces a shallow tree. It groups tokens into NP chunks
("a US patent case", "its rival") and PP chunks ("in August", "to Apple").
The coordination "lost … and was ordered" is not captured because the hand-written
grammar has no rule for coordinated VPs. Many tokens remain outside any phrase.

**spaCy dependency output:**
Every token is linked to exactly one head.
"Samsung" is the nsubj of "lost" (root). "case" is the dobj.
"Apple" is the pobj of the preposition "to". "ordered" is joined via conj.
The full sentence structure is captured, including nested PPs and the conjunction.

**Key difference:** NLTK produces a partial, flat chunking limited by the rules we write;
spaCy produces a complete word-level dependency graph for the entire sentence.
